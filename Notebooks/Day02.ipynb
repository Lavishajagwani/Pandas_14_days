### 2.4 Find the items with price between 100 and 200 (inclusive) using between(). Then find the items outside that range using ~. Check that the two counts add up to the total number of items.

```python
condition = items.price.between(100, 200, inclusive='both')

a = items[condition].shape
b = items[~condition].shape
print(a)
print(b)
```

```python
c = items.shape
print(c)
```

## Task 3 - Use apply once for a custom function, then rewrite it vectorised

### 3.1 Write a function size_label(w) that returns 'light' below 500 g, 'medium' below 5000 g and 'heavy' otherwise. Use apply to fill product_size from product_weight_g. What label do the missing weights get, and is that right?

```python
def size_label(w):
    ## Changes made for handling missing weights
    if pd.isna(w):
        return'unknown'
    elif w < 500:
        return 'light'
    elif w < 5000:
        return 'medium'
    return 'heavy'
    
```

```python
products.head()
```

```python
%%timeit
products["product_size"] = products.product_weight_g.apply(size_label)
```

```python
products.head()
```

```python
products[products.product_weight_g.isnull()]
```

### 3.2 Rewrite it without apply, using np.select with a list of conditions and a list of labels. Time both versions with %%timeit.

```python
import numpy as np
conditions = [
    ## Changes made for handling missing weights
    products.product_weight_g.isnull(),
    products.product_weight_g < 500,
    products.product_weight_g < 5000
]

choices = ['unknown','light','medium']
```

```python
%%timeit
products["product_size_2"] = np.select(conditions,choices,default="heavy")
```

```python
products.head()
```

```python
products[products.product_weight_g.isnull()]
```

Apply function calls the function once per row (about 33,000 times), and each call is ordinary Python, which is slow.

np.select takes the whole column at once, and works out all the labels in one pass inside NumPy. 
How it works: for each row, np.select goes through the conditions in order and uses the label of the first one that is True. If none are True, it uses default.

Here the order for the label matters: w < 5000 is also True for a light product, but the first match wins, so light is used.

Q. In both the ways - Apply() and np.select, product_weight value in each row is checked, so how does the apply function takes more time than np.select?

A. You're right that both versions look at every weight. The difference is who does the looking, and how much work surrounds each check.

apply: the loop runs in Python

For each of the roughly 33,000 rows, pandas has to:

take the value out of the column and wrap it as a Python object,
call your Python function,
run the if / elif / else lines through the Python interpreter, which checks types and looks up names as it goes,
collect the returned string into a list.

Each step is cheap, but repeated 33,000 times the overhead adds up. Python is flexible, and that flexibility costs time on every operation.

np.select: the loop runs in compiled code

w < 500 is a single operation on the whole column. NumPy stores the weights as one block of raw numbers in memory and compares them all in a tight loop written in C. There is no function call per row, no Python objects created, and the processor can handle many numbers at once. Then w < 5000 does the same for the second condition, and np.select combines the results, again in compiled code.

An analogy

Imagine marking 33,000 exam papers.

apply is like handing each paper to a clerk one at a time, who reads the marking rules from a book, decides the grade, and writes it down before taking the next paper.
np.select is like running all the papers through a machine that stamps every paper at once.

Both give each paper a grade, but one does a great deal of setup work per paper.

Even though np.select checks more

It actually scans the column once for each condition, so it does more raw comparisons than apply, which stops at the first if that is true. It is still much faster, because each comparison is so cheap in compiled code that doing two or three passes costs less than one pass of Python function calls.

The general rule for pandas

Operations that work on whole columns (comparisons, arithmetic, np.select, np.where, .str methods) are called vectorised, and they run in compiled code. apply, iterrows and Python loops run row by row in Python. The difference is usually large and grows as the table gets bigger, which is why Day 12 includes the iterrows versus vectorised comparison.

apply is still fine for small tables, or for logic too complicated to express with column operations. It's just not the first choice when a vectorised version exists.

## Excercise - In products and category_name_translation, normalise category names, translate Portuguese to English, and list the 10 categories with the most products.

```python
category_translation.head()
```

```python
products.head()
```

```python
prod = products.copy()
ct = category_translation.copy()

# 1. Normalise: strip spaces and lowercase, in both tables
prod["product_category_name"] = prod["product_category_name"].str.strip().str.lower()
ct["product_category_name"] = ct["product_category_name"].str.strip().str.lower()
ct["product_category_name_english"] = ct["product_category_name_english"].str.strip().str.lower()

# 2. Build the Portuguese -> English lookup
mapping = dict(zip(ct["product_category_name"], ct["product_category_name_english"]))

# 3. Translate
prod["category_en"] = prod["product_category_name"].map(mapping)

# 4. Handle the leftovers (see below)
prod["category_en"] = (prod["category_en"]
                       .fillna(prod["product_category_name"])   # untranslated: keep Portuguese name
                       .fillna("unknown"))                       # no category at all

# 5. Top 10 categories by number of products
top10 = prod["category_en"].value_counts().head(10)
top10
```

Handle two kinds of missing values. 
After map, category_en has NaNs for two different reasons:

1. The product has no category at all (about 600 products), and
2. The product has a category that isn't in the translation table (a couple of categories).

The first fillna keeps the Portuguese name for untranslated categories, so they aren't mixed up with products that have no category. The second fills the true blanks with "unknown"

```python
##Check for null category
products["product_category_name"].isna().sum()                                   # no category
```

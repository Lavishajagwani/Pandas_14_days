## Task 2 - Use map and replace with a dictionary to standardise categories

### 2.1 In payments, create payment_label with map() and a dictionary: credit_card to 'Credit card', boleto to 'Bank slip', voucher to 'Voucher', debit_card to 'Debit card'. What happened to 'not_defined', and why?

'not_defined' became NaN, because it isn’t a key in the dictionary.

```python
payments.head()
```

```python
payments.payment_type.value_counts()
```

```python
mapping = {"credit_card" : "Credit card",
           "boleto" : "Bank slip",
           "voucher" : "Voucher",
           "debit_card" : "Debit card"}

payments["payment_label"] = payments.payment_type.map(mapping)
```

```python
payments.payment_label.value_counts()
```

```python
payments.payment_label.isna().sum()
```

```python
payments[payments.payment_label.isna()]
```

### 2.2 Do the same with replace(). How does it treat 'not_defined' differently?

replace() leaves values that are not in the dictionary unchanged. Use map when you want unmatched values to show up as NaN, and replace when you want them kept.

```python
mapping = {"credit_card" : "Credit card",
           "boleto" : "Bank slip",
           "voucher" : "Voucher",
           "debit_card" : "Debit card"}

payments["payment_label_2"] = payments.payment_type.replace(mapping)
```

```python
payments.payment_label_2.value_counts()
```

```python
payments.payment_label_2.isna().sum()
```

```python
payments[payments.payment_label_2.isna()]
```

```python
payments[payments.payment_label.isna()]
```

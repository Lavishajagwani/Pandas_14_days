## Task 2 - Choose between dropna, fillna, and leaving nulls alone, and write down why for each column


### 2.1 For each of these columns, decide whether to drop the rows, fill the nulls, or leave them, and write one sentence on why: orders.order_delivered_customer_date, products.product_category_name, products.product_weight_g, reviews.review_comment_message.

### My Answer
Nulls in orders.order_delivered_customer_date should be left as it is, since it is timestamp column.

Nulls in products.product_category_nme can be filled with "Unknown" 

Nulls in products.product_weight_g can be filled with the mean weight of the column

Nulls in reviews.review_comment_message can be left as it is.

### Right answer
1. order_delivered_customer_date: leave it. The decision is right, but “it is a timestamp column” isn’t the reason. The type doesn’t matter. The reason is what the null means: the order hasn’t been delivered, was canceled, or is still in transit. A null here is real information, and filling it would invent a delivery that never happened. Also don’t drop these rows from orders, because that would remove canceled and undelivered orders from your order counts. Drop them only inside a calculation that needs a delivery date, such as delivery time.

2. product_category_name: fill with “unknown”. This is right. You keep the product rows and make the missing category an explicit group, so it shows up in group counts instead of disappearing. Use lowercase 'unknown' so it matches the style of the other category names.

3. product_weight_g: use the median, not the mean. Filling is the right idea, since only a couple of products are missing a weight and dropping them would waste their other data. The median is the safer fill, because a few very heavy products pull the mean up, while the median stays near a typical product. With only about 2 missing values the practical effect is tiny, but the median is the better habit. A step beyond that is the median weight within the same product category, since a book and a sofa don’t weigh the same.

4. review_comment_message: leave it. Correct. A null means the customer left no comment. That is not a data error, and many reviews have only a score. Filling with something like “no comment” would add fake text that shows up in word counts and searches. When you analyze comments, use na=False in string filters (such as .str.contains('x', na=False)) so the nulls don’t cause errors.

A pattern for these decisions: ask what the null means. If it means “hasn’t happened” or “was not given”, leave it. If it means “wasn’t recorded” for something every row should have, fill it with a placeholder or a typical value. If the row is useless without that value, drop it.

### Exercise - Load all 9 Olist tables into a dictionary of DataFrames. Print the shape and null count of each. Write down which columns look like keys.

Path(...) turns the folder string into a path object, which has useful methods like glob. It also handles Windows backslashes and joining paths cleanly.

glob("*.csv") searches that folder for files whose names match a pattern. * means “anything”, so *.csv matches every file ending in .csv. It returns the matching file paths one by one. Other patterns work too:

glob("olist_*"): files starting with olist_
glob("*order*.csv"): CSV files with “order” in the name
rglob("*.csv"): also searches inside sub-folders

f.stem is the file name without the folder and without the extension. For C:\...\Data\olist_orders_dataset.csv:

f.name gives olist_orders_dataset.csv
f.stem gives olist_orders_dataset
f.suffix gives .csv

The dictionary comprehension {f.stem: pd.read_csv(f) for f in ...} loops over every CSV file, reads it into a DataFrame, and stores it under its stem. The result is a dictionary, with file names as keys and DataFrames as values:

```python
## Load all the 9 tables together

from pathlib import Path

data_dir = Path(r"C:\Users\DELL\Desktop\Pandas_in_14_days\Work\Data")
tables = {f.stem: pd.read_csv(f) for f in data_dir.glob("*.csv")}

for name, df in tables.items():
    print(name, df.shape)
```

```python
## Print count of null values in each column of a dataframe

for key in tables.keys():
    count_of_nulls = tables[key].isna().sum()
    print(count_of_nulls)
```

```python
## Print count of null values in each column of a dataframe

for key, value in tables.items():
    print(key, value.shape)
    print(value.isna().sum())
```

```python
## Print count of null values in total in a dataframe (adding up from all the columns)

for key, value in tables.items():
    print(key, value.shape)
    print(value.isna().sum().sum())
```
